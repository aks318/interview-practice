```python
from langgraph.graph import StateGraph, START, END, MessagesState
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
```

```python
from typing import TypedDict


llm = ChatOllama(model="llama3.2:1b")

class State(TypedDict):
    BE_Team: str
    FE_Team: str
    Testing_Team: str
    Review: str

```

```python
def manager(state):
    print("manager recevied the requirement")
def BE_Team(state):
    return {"BE_Team" : "BE task completed"}
def FE_Team(state):
    return {"FE_Team" : "FE task completed"}
def Testing_Team(state):
    return {"Testing_Team" : "Testing task completed"}
def Review(state):
    return{"Review" : "Review Completed"}
```

```python
graph = StateGraph(State)

graph.add_node("manager", manager)
graph.add_node("BE_Team", BE_Team)
graph.add_node("FE_Team", FE_Team)
graph.add_node("Testing_Team", Testing_Team)
graph.add_node("Review", Review)

graph.add_edge(START, "manager")

graph.add_edge("manager", "BE_Team")
graph.add_edge("manager", "FE_Team")
graph.add_edge("manager", "Testing_Team")

graph.add_edge("BE_Team", "Review")
graph.add_edge("FE_Team", "Review")
graph.add_edge("Testing_Team", "Review")

graph.add_edge("Review", END)

workflow = graph.compile()

workflow.invoke({})



```
