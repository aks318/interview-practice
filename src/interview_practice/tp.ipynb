```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate

from pydantic import BaseModel

llm = ChatOllama(model= "llama3.2:1b")
```

```python
class ModelOutput(BaseModel):
    age: int
    name: str
    date: str

llm_with_structure_outout = llm.with_structured_output(ModelOutput)

prompt = PromptTemplate(template="""
    I am aakash, I was born on November 3rd, 1998. Can u tell me my age,
    and can u tell me what is the todays date?
""")

chain = prompt | llm_with_structure_outout

chain.invoke({})
```
